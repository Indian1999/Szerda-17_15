# Tanító szöveg kialakítása

> Egy nagy stringben az összes szó





```python
data = pd.read_csv("EcoPreprocessed.csv")
text = data["review"]
```

```python
tokenizer = Tokenizer()
tokenizer.fit_on_texts(text.values)

text_tokens = tokenizer.texts_to_sequences(text.values)
all_tokens = []
for item in text_tokens: # [31, 343, 321, 664, 123, 12]
    for token in item:
        all_tokens.append(token)

vocab_size = len(tokenizer.word_index)
print("Vocab_size:", vocab_size)
print(all_tokens[:20])
```

```python
sentence_len = 15
pred_len = 1
input_len = sentence_len - pred_len
x = []
y = []
for i in range(len(all_tokens) - sentence_len):
    x.append(all_tokens[i:i+input_len]) # 14
    y.append(all_tokens[i+input_len:i+sentence_len])

print(x[:2])
```
